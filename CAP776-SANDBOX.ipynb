```python
#FACTORIAL
def fact(x):
    if x==0:
        return 1
    else:
        return x * fact(x-1)
print(fact(5))
```

```python
#TOWER OF HANOI
def move(f,t):
    print(f"From {f} to {t}")

def hanoi(n,f,h,t):
    if n==0:
        pass
    else:
        hanoi(n-1,f,t,h)
        move(f,t)
        hanoi(n-1,h,f,t)
hanoi(4,"A","B","C")
```

Employee Attendance Checker
An employee's attendance for seven days is stored as values such as "P" for present and "A" for absent. Write a function that counts the number of present and absent days using a loop. If the employee was present for at least 6 days, display "Excellent Attendance". If present for 4 or 5 days, display "Satisfactory Attendance". Otherwise display "Poor Attendance".
Password Attempt System

```python
def attendance_calculator(numbers):
    presentdays=0
    for i in numbers:
        if i=="P":
            presentdays+=1
    if presentdays>=6:
        print("Excellent Attendance")
    elif presentdays==4 or presentdays==5:
        print("Satisfactory Attendance")
    else:
        print("Poor attendance")
attendance=["P","P","A","A","P","P","P"]
attendance_calculator(attendance)
```

Bus Fare Calculator
Write a function calculate_fare(age, distance) that calculates bus fare. The normal fare is ₹5 per kilometre. Children below 12 receive a 50% discount, senior citizens aged 60 or above receive a 40% discount, and all others pay the full fare. The function should return the final fare.

```python
def calculate_fare(age,distance):
    fare=distance*5
    if(age<12):
        fare=fare*0.5
        return fare
    elif(age>=60):
        fare=fare*0.4
        return fare
    else:
        return fare
age=int(input("enter age"))
distance=float(input("enter distance"))
final_fare=calculate_fare(age,distance)
print(f'Final fare ={final_fare}')
```

Dictionary Questions
A school maintains the marks of students in a Python dictionary, where the student name is the key and the marks obtained are the value.
marks = {
    "Aarav": 78,
    "Diya": 92,
    "Kabir": 65,
    "Meera": 88,
    "Rohan": 72
}
Answer the following questions based on the above dictionary:
a) Write a statement to display the marks obtained by Meera.
b) Write a statement to add a new student "Ananya" with 85 marks.
c) Write a statement to change Kabir's marks from 65 to 75.
d) Write Python code to find and display the student who scored the highest marks.

```python
student_marks={
    "Aarav": 78,
    "Diya": 92,
    "Kabir": 65,
    "Meera": 88,
    "Rohan": 72
}
print(f'Marks obtained by Meera are {student_marks["Meera"]}')
student_marks["Ananya"]=85
student_marks["Kabir"]=75
print(f'Marks obtained by Kabir are {student_marks["Kabir"]}')
max_marks=0
topper_student=""
for student in student_marks:
    if student_marks[student] > max_marks:
        max_marks = student_marks[student]
        topper_student = student
print(f'Student with max marks is {topper_student}')
```
