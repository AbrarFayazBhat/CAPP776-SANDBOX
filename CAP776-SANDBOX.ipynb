```python
#create a function to calculate the expenses of morning,evening and default morning value is 10:am
def calculate_expenses(morning,afternoon,evening):
    print(f"Morning expenses: {morning}")
    print(f"afternoon expenses: {afternoon}")
    print(f"evening expenses: {evening}")
    print(f"total expenses: {morning+afternoon+evening}")
calculate_expenses(100,200,400)
    
```

```python
def fact(x):
    if x==0:
        return 1
    else:
        return x * fact(x-1)
print(fact(5))
```
